```python
import requests
from typing import Dict, Any, Tuple

GEOCODING_URL = "https://geocoding-api.open-meteo.com/v1/search"
FORECAST_URL = "https://api.open-meteo.com/v1/forecast"

def get_coordinates(location_name: str) -> Tuple[float, float, str]:
    """Resolves a city name to latitude, longitude, and formatted display name.
    
    Args:
        location_name: Name of the target city or region.
        
    Returns:
        Tuple containing (latitude, longitude, display_name).
        
    Raises:
        ValueError: If location cannot be found.
    """
    params = {"name": location_name, "count": 1, "language": "en", "format": "json"}
    response = requests.get(GEOCODING_URL, params=params, timeout=10)
    response.raise_for_status()
    
    data = response.json()
    if not data.get("results"):
        raise ValueError(f"Location '{location_name}' not found.")
        
    result = data["results"][0]
    display_name = f"{result['name']}, {result.get('country', '')}"
    return float(result["latitude"]), float(result["longitude"]), display_name


def fetch_weather_forecast(lat: float, lon: float, days: int = 5) -> Dict[str, Any]:
    """Fetches daily weather forecast data for given coordinates.
    
    Args:
        lat: Latitude coordinate.
        lon: Longitude coordinate.
        days: Number of forecast days (1-16).
        
    Returns:
        Dictionary containing weather forecast data.
    """
    params = {
        "latitude": lat,
        "longitude": lon,
        "forecast_days": days,
        "daily": [
            "temperature_2m_max",
            "temperature_2m_min",
            "precipitation_sum",
            "wind_speed_10m_max"
        ],
        "timezone": "auto"
    }
    response = requests.get(FORECAST_URL, params=params, timeout=10)
    response.raise_for_status()
    
    return response.json()
```
